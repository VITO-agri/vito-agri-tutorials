Execute the following cell to unzip a folder.<br>

Before running the cell, make sure that this notebook is located in the same folder containing the .zip file<br>

Upon executing the cell, you will be prompted to provide the name of the zip file you wish to unzip.<br>
Do NOT include the extension (.zip)!!

```python
import zipfile
from pathlib import Path

# Get path to zip file
zipf = input("Enter the name of the zip file: ")
full_path = Path(f'./{zipf}.zip')
if not full_path.exists():
    raise FileNotFoundError(f"File {full_path} not found")

# Create output directory
Path(f'./{zipf}').mkdir(exist_ok=True)

# Extract files
with zipfile.ZipFile(str(full_path), 'r') as zip_ref:
    zip_ref.extractall(f'./{zipf}')
    print(f"Files extracted to {full_path.parent}/{zipf}")
```
